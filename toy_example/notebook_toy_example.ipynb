# Toy Example: True vs. Data-Estimated vs. Trained-Model Lipschitz Constant

## Motivation

A model's **Lipschitz constant** `L` bounds how much its output can change in response to a small change in its input: `|f(x') - f(x)| <= L * ||x' - x||` for all `x, x'`. This is one of the tools used to reason about a neural network's sensitivity to input perturbations (e.g. adversarial examples). A smaller Lipschitz constant means a smoother, more robust function whereas a larger constant means function is more sensitive to small changes (i.e. it is more "stretchy"). In practice we never have the true function `f*`, only a trained model `f_hat` fit to a finite sample of it, so any Lipschitz constant we compute is really an *estimate*, and its quality depends on both how much data we had and where that data was located.

This experiment builds a **toy regression problem where the true function `f*` and its true Lipschitz constant `L*` are known essentially exactly**, so we can directly measure how far off our estimates are — something that's computationally infeasible to check on a real dataset like MNIST, where `L*` is unknown. This Toy example serves as a simple example validating out methodlogoly going forward. 

## The three quantities being compared

1. **`L*`** — the *true* Lipschitz constant of the ground-truth function.
   - **Tier A** (a single `A * tanh(w^T x + b)` ridge): closed-form, `L* = A * ||w||`.
   - **Tier B** (a sum of 2-3 such ridges): no closed form, so it's estimated numerically — dense grid search over the domain, refined by gradient ascent (`torch.optim.LBFGS`) on `||grad f*(x)||` from many random restarts. This is treated as "true" for the rest of the experiment because it is exact to numerical precision.
2. **`L_hat_data`** — a *data-only* estimate: given only the sampled `(x_i, f*(x_i))` pairs (no model), take the steepest pairwise slope, `max_{i != j} |y_i - y_j| / ||x_i - x_j||`.
3. **`L_hat_model`** — estimates computed from a model `f_hat` *trained* on that sample, via three different sub-methods:
   - **pairwise** — the same steepest-slope formula as `L_hat_data`, but applied to the model's predictions instead of the raw data (so it can be evaluated anywhere, e.g. on points with no training data nearby).
   - **local-perturbation** — a finite-difference estimate: sample many points in a small ball around a query point `x0`, take the steepest slope from `x0` to any of them.
   - **gradient-norm** — the infinitesimal/exact-derivative estimate: `||grad f_hat(x0)||` via autograd. This is the local Lipschitz constant of the model's tangent plane at `x0`.

Both `L_hat_data` and `L_hat_model` can also be computed with a **Mahalanobis distance** derived from a polynomial embedding of `x`, instead of plain Euclidean distance — see below and README.md's "Distance metric: Euclidean vs. Mahalanobis" section.

## What the experiment is designed to reveal

Because `L*` is known exactly here, we can ask questions that are normally unanswerable: 
- Does `L_hat_model` converge to `L*` as training data grows?
- Does model capacity matter?
- A key point of this experiment — what happens to `L_hat_model` in a region where training data was deliberately **sparse** (a "gap")? Does the trained model *flatten out* there (underestimate `L*`), or does it *oscillate* and overshoot it? This notebook builds that gap deliberately and measures the effect directly.

## Step 3: Main experiment — gap vs. uniform sampling (Tier B, d=1)

Now the real question. `run_main_experiment` does the following:

1. Builds a **Tier B ground truth**: a sum of three `tanh` ridges of different steepness (`||w||` = 8, 2, and 5) placed at different offsets `b`, so they don't overlap. There is no closed form for this function's Lipschitz constant, so `L*` and the point `x*` where it is attained are found via the grid-search + gradient-ascent procedure in `tier_b_true_L`.

2. Builds **two datasets of the same size `N`**: a `uniform` dataset (ordinary i.i.d. sampling over the whole domain) and a `gap` dataset, which is identical except that the region around `x*` — precisely where the function is steepest — is deliberately undersampled (only a small `gap_fraction` of points fall there).

3. Trains a `TinyMLP` on each dataset independently.

4. For each trained model, evaluates `L_hat_data` (from the raw training sample), `L_hat_model` pairwise (on the training points and on a dense grid spanning the whole domain, including inside the gap), and the local-perturbation and gradient-norm estimates *at every point on that grid* — not just their max, so we can see the whole shape.

The resulting plot overlays `f*(x)`, the trained `f_hat(x)`, and the pointwise local-Lipschitz (finite-difference) curve for both datasets side by side, with the training points shown as a rug plot along the x-axis so you can see exactly where each model did and didn't get to see data. 

**What to look for:** compare the height of the red local-Lipschitz curve near `x*` between the two panels — if the gap dataset's peak is visibly lower than the uniform dataset's, that is the model *flattening out* (underestimating `L*`) precisely where its training data was sparse.

```python
main_results = run_experiment.run_main_experiment()
```

### Quantifying the gap effect

`report` below lists, per dataset, the max local-Lipschitz estimate inside the gap region vs. outside it — a lower in-gap max for the gap dataset relative to the uniform dataset is the numeric confirmation of the flattening effect described above (rather than just eyeballing the plot).

```python
for dataset_key, r in main_results["report"].items():
    print(dataset_key, r,'\n')
```

## Step 4: Sweeps — does more data or more capacity fix the gap?

Two sweeps, each repeated for both the `uniform` and `gap` datasets:

- **N-sweep**: fix model capacity, vary training-set size `N ∈ {50, 100, 200, 500, 1000, 2000, 5000}`.
- **Capacity-sweep**: fix `N`, vary the `TinyMLP` hidden width `∈ {4, 8, 16, 32, 64, 128}`.

In both, `L_hat_model` is evaluated on a **fixed, large, held-out grid** — not on the training points — specifically so the numbers reflect how well the model *generalizes* to steep regions it may not have trained on, not how well it merely memorized its own training set.

Each plot shows `L*` as a horizontal reference line alongside the `L_hat_data` / `L_hat_model` curves. 

**What to look for:** on the `uniform` dataset, both curves should climb toward `L*` as N or width increases. On the `gap` dataset, watch whether `L_hat_model` climbs toward `L*` as capacity increases at fixed `N` — if it stays stuck below `L*` across the whole width range, that's evidence that the information genuinely isn't there to recover (a coverage problem, not a capacity problem).

```python
sweep_results = run_experiment.run_sweeps()
```

## Step 5: Seed-averaged gap N-sweep — is the non-monotonicity real?

The gap dataset's N-sweep plot above: `L_hat_model` vs. `N` is **non-monotonic** — it bounces between `~4.6` and `~8.4` as `N` grows from 50 to 5000, unlike the uniform dataset's clean climb toward `L*`, at the single seed this notebook otherwise uses throughout. `run_gap_N_sweep_seed_averaged` tests whether this pattern holds across seeds or is single-seed noise.

It repeats the identical gap-dataset N-sweep across `n_seeds` independent seeds — varying both the sampled gap dataset and the model's initialization/training at every `N`, with every other hyperparameter held fixed — and reports the mean, spread, and every individual seed's trajectory. This is slower than a single-seed sweep (5x the model-training cost by default) and is not part of `run_sweeps()` or `main()` — an opt-in check, run here on its own.

```python
seed_avg_results = run_experiment.run_gap_N_sweep_seed_averaged(n_seeds=5)
```

**Result:** the bounce is mostly single-seed noise — individual-seed curves (faint lines in the plot above) scatter widely at every `N`, and no two seeds trace the same shape. Averaging across seeds does not reveal convergence toward `L*` either: the mean `L_hat_model` curve stays essentially **flat** at `~6.3-7.2` across the *entire* `N` range, never climbing toward `L*=8.46` the way the mean `L_hat_data` (data-only, no model) does over the same range. At the largest `N` tested, `L_hat_model = 6.89 ± 0.56` (mean ± std across 5 seeds) vs. `L*=8.46`.

More gap-sampled data does not resolve the undershoot, the same pattern the capacity sweep below shows for model width.


## Step 6: 2D extension — the version that generalizes to MNIST

Everything above is 1D, which is easy to plot as a curve but not representative of real high-dimensional inputs. This step repeats the Tier B setup in `d=2`, with three ridges pointing in different directions (steep along `x1`, shallow along `x2`, moderate diagonally), trains a `TinyMLP` on a 2D dataset with the same kind of deliberate sampling gap, and produces three Lipschitz heatmaps over `[-5,5]^2`:

1. the **true** `||grad f*(x)||` (ground truth, from the analytic gradient),
2. the **model's** `||grad f_hat(x)||` (autograd, the gradient-norm estimator),
3. the **finite-difference local Lipschitz estimate** (the local-perturbation estimator),

with the training points scattered on top of each so gap regions and "hot spots" in the estimates can be visually compared directly.

A fourth output, saved separately, is a **coverage-density diagnostic** (`local_sample_density` -> `plot_coverage_heatmap`,
`results/step6_coverage_heatmap.png`): for each grid point, how many training points actually fell within `local_radius` of it. This isn't a Lipschitz quantity itself — it answers a different question (was this region tested at all?).

```python
results_2d = run_experiment.run_2d_extension()
```

```python
results_2d["figure"]
```

**Coverage check**: at the settings used above (`gap_radius=0.7`, `gap_fraction=0.03`), the ridge-intersection hotspot at `x*` is **not** undersampled — its local density (3, at `x*`) is actually *above* the grid-wide average (~2.15), because `gap_fraction=0.03` is nearly 2x the ~0.0154 density a plain uniform sample would put in a ball that size by area alone.

Correcting to `gap_fraction≈0.0077` (half that naive-uniform baseline — `run_experiment.run_2d_extension(gap_fraction=0.0077)`) produces genuine local undersampling instead (density 0 vs. grid-wide mean ~2.15). Averaged over 5 seeds (varying both the sampled dataset and the model's initialization), the hotspot's local-Lipschitz estimate then undershoots the true gradient norm by ~9% on average (mean ratio 0.91, stdev 0.07) — real, but far weaker than the 20-50%+ undershoot seen in the 1D case above. Why the 2D effect is weaker is not established; one hypothesis is that a 2D gap can be approached from more directions than a 1D one, giving a smooth model more surrounding signal to interpolate the peak from.

## Step 7: Does the choice of distance metric matter?

Every estimator used above measures how "close" two input points are using plain Euclidean distance in raw `x`, though nothing about the estimators forces that choice. This step tests an alternative: embed `x` into a richer feature space `(x, x^2, ..., x^degree)`, derive a Mahalanobis distance from the empirical covariance of the embedded training points, and use that distance in place of Euclidean distance inside the same estimators (`pairwise_lipschitz`, `local_perturbation_lipschitz` now both accept an optional `embed_fn`/`precision` pair for exactly this).

`run_metric_embedding_check` reuses the Tier A gap-sampled dataset (single ridge, closed-form `L*`) and computes `L_hat_data` twice on the *same* raw data — once with plain Euclidean distance, once with the Mahalanobis distance — so the only thing that changes between the two numbers below is the metric.

```python
metric_check_results = run_experiment.run_metric_embedding_check()
```

**Result**: `L_hat_plain` undershoots `L*=6.0` by about 19% (`~4.87`);
`L_hat_mahalanobis`, computed from the identical raw data, lands within
1% of `L*` (`~6.01`). The metric alone accounts for that gap.

### Choosing the embedding degree

The polynomial degree is a free parameter, and it matters more than it
looks: too low and the embedding has too little structure to help; too
high and the embedded covariance becomes ill-conditioned (polynomial
powers are highly collinear on a bounded domain), making the resulting
metric numerically unreliable even though it still technically inverts.
`sweep_polynomial_degree` checks both accuracy against `L*` and the
condition number of the covariance, degree by degree, and picks the
lowest degree that is both accurate and well-conditioned rather than
whichever looks best on accuracy alone.

```python
degree_sweep_results = run_experiment.sweep_polynomial_degree()
```

**Result**: degree 3 is the clear choice — relative error against `L*`
drops to ~0.08% (from >130% at degree 1-2), while the condition number of
the covariance (`cond(cov)`) stays at a modest ~1.5e3. Degrees 5-6 add
more features, but the condition number blows up past 1e6 and accuracy
collapses along with it — more free parameters here buys overfitting to
sampling noise, not a better metric.

### From a single global number to a local curve

The comparison above is a single scalar for the whole domain. The more
useful version is **local**: a Lipschitz estimate at every point `x`, so
we can see *where* the plain-Euclidean estimate falls short rather than
only that it does on average. `run_tier_a_gap_demo` (the same call already
made in the Tier A gap demo above) now computes this automatically —
alongside the original plain-Euclidean local-Lipschitz curve, it computes
the Mahalanobis-distance version and the Mahalanobis global scalar, using
the degree-3 embedding just justified above. No need to retrain anything;
`tier_a_gap_demo_results` from that earlier cell already has it.

**What to look for:** the two dotted horizontal lines are the global
scalars (plain vs. Mahalanobis) — same story as `run_metric_embedding_check`
above, now visible directly against the true `L*` line. The two solid
curves are the local estimates, plain and Mahalanobis, which should both
peak sharply at `x*` and decay to ~0 away from it, tracking each other
closely (the local estimate doesn't need "help" from a better metric to
find the peak here — it's the *global* scalar that benefits most from the
metric change).

`f(x)` itself can also be appended as an extra embedding coordinate
(`embeddings.py::augmented_embedding`), on the reasoning that a metric
aware of the function's own shape could weigh a disturbance in an
already-flat region more heavily than the same-size disturbance somewhere
already changing fast. This is self-cancelling in practice: a metric
built from `f*` to measure `f*`'s own Lipschitz behavior divides the
signal by almost exactly itself, and the flattening effect this whole
notebook is about becomes invisible rather than clearer. Only the plain
polynomial embedding is used here.

## Step 8: Summary and status

Running this notebook end-to-end (with the fixed seed used throughout) reproduces the four central findings of this experiment:

- **The gap causes flattening, not oscillation.** In Step 3, the trained
  model's local-Lipschitz estimate near `x*` is visibly *lower* for the
  gap-sampled dataset than for the uniformly-sampled one — the model
  underestimates the true steepness where it lacked data, rather than
  overshooting it.
- **Capacity does not fix a coverage gap.** In Step 4's capacity-sweep on
  the gap dataset, `L_hat_model` stays below both `L_hat_data` and `L*`
  across the entire range of hidden widths tested (4 to 128) — more
  parameters alone cannot recover information that was never in the
  training sample.
- **The distance metric used to measure closeness matters too.** In
  Step 7, switching only the distance metric — from plain Euclidean to a
  Mahalanobis distance derived from a polynomial embedding of `x`, on the
  exact same raw data — cuts the error of the data-only estimate against
  `L*` from ~19% to under 1%. Getting this benefit reliably depends on
  picking a well-conditioned embedding degree (Step 7 also shows why
  degree 3 was chosen here).
- **The flattening effect is much weaker in 2D, and only shows up once
  the gap is calibrated correctly.** In Step 6, the *default* sampling
  parameters don't actually undersample the ridge-intersection hotspot;
  once corrected so they do, the hotspot's local-Lipschitz estimate
  undershoots the true value by only ~9% on average across 5 seeds —
  real, but far below the 20-50%+ undershoot in the 1D case. Why is not
  yet established.
